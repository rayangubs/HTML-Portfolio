##CHATBOTS

Natural Language Processing (NLP) chatbots are conversational systems that use NLP to process user input and provide relevant responses. They can communicate through text or voice and are commonly used in customer service, education, virtual assistance and other applications.

**Types of Chatbots**

NLP chatbots can be classified based on how they process user input, select or generate responses and use conversation context.

- **Rule-Based Chatbots**: Follow predefined rules, keywords and response patterns to handle user queries.
- **Retrieval-Based Chatbots**: Select the most appropriate response from a predefined set based on the user's input or intent.
- **Generative Chatbots**: Generate new responses using machine learning or large language models (LLMs).
- **Hybrid Chatbots**: Combine rule-based approaches with machine learning or NLP techniques.
- **Contextual Chatbots**: Use conversation history and context to provide more relevant responses across multiple interactions.

<b>How Chatbots work?</b>

image.png

NLP chatbots typically follow these steps:

- <b>Input Processing</b>: The chatbot processes the user's text or speech using techniques such as tokenization and named entity recognition.
- <b>Intent Recognition</b>: It identifies the user's intent, such as asking a question, requesting information or performing an action.
- <b>Dialogue Management</b>: The chatbot maintains the conversation context and determines what should happen next.
- <b>Response Generation</b>: It retrieves or generates an appropriate response using a knowledge base, rules or a language model.
- <b>Response Delivery</b>: The generated response is presented to the user as text or speech.
- <b>Learning and Improvement</b>: AI-based chatbots can use interaction data and feedback to improve their performance.

NLTK includes a built-in **chat** module that provides tools for creating simple rule-based chatbots. One important feature is the **reflections** dictionary, which helps a chatbot transform pronouns and related words in a user's message.

For example:
- "I am" → "you are"
- "my" → "your"
- "I" → "you"
- "you" → "I"


```python
import nltk
from nltk.chat.util import Chat, reflections

# Define some simple rules-based responses
rules = [
    (r'(.*)hello(.*)', ['Hello!', 'Hi there!', 'Hey!']),
    (r'(.*)how are you(.*)', ['I am doing well, thank you!', 'I\'m great, thanks for asking!']),
    (r'(.*)What is your name?(.*)', ['I am an NLP chatbot.', 'You can call me ChatBot.']),
    (r'(.*)exit(.*)', ['Goodbye!', 'Bye!', 'Take care!']),
    (r'(.*)what is my name?(.*)', ['Your name is Rayan', 'You are Rayan', 'You are poopy']),
]

# Create a chatbot instance
chatbot = Chat(rules, reflections)

# Start chatting with the bot
print('Welcome to the NLP ChatBot. Type \'exit\' to end the conversation')
while True:
  user_input = input('You: ')
  response = chatbot.respond(user_input.lower())
  print('Bot:', response)
  if user_input.lower() == 'exit':
    break
```

##What is a Retrieval-Based Chatbot?
A retrieval-based chatbot is a chatbot that answers users by finding the most relevant response from a predefined collection of information, such as FAQs, product details, or stored answers.

Instead of generating completely new answers, it selects an existing answer based on the user's question.

##How it works?

image.png

<B>How does it work?</B>
1. Input: The user enters a question.
2. Preprocessing: The system processes the question into a form suitable for comparison.
3. Vectorization: The question is converted into numerical features or a vector.
4. Retrieval: The system compares the question against stored questions or documents.
5. Similarity calculation: It measures how closely the user's question matches the stored information.
6. Response selection: The chatbot returns the answer associated with the best match, provided the match is sufficiently relevant.

Common retrieval methods include keyword matching, TF-IDF, cosine similarity, and semantic embeddings.

## Common libraries used in Retrieval based chatbots

- **TfidfVectorizer**: Converts text into numerical vectors so the computer can compare questions.

- **cosine_similarity**: Measures how similar the user's question is to the stored questions.


<B>2. What is cosine similarity?</B>

Cosine similarity is a mathematical method used to measure how similar two vectors are by calculating the cosine of the angle between them.

In a retrieval-based chatbot, it can measure how similar the user's question vector is to the vectors representing stored questions or documents.

The result is commonly interpreted as follows when using standard, nonnegative TF-IDF vectors:

- 1.0 — identical direction, indicating very high similarity.
- 0.0 — no overlap in vector dimensions.
- A value between 0.0 and 1.0 — varying degrees of similarity.



Cosine similarity compares the user's question vector with each stored question vector.

#Applying cosine similarity to text using Bag of Words

```python
#Define documents
documents = [
    'machine learning is fun',
    'machine learning is powerful',
    'the weather is sunny'
]

#Convert documents into vectors

from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(documents)

# Get feature (word) names
words = vectorizer.get_feature_names_out()

print('Vocabulary:')
print(vectorizer.get_feature_names_out())

print('\nBag of Words Matrix:')
print(bow_matrix.toarray())

#Calculate cosine similarity

from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(bow_matrix)

print('\n', similarity_matrix.round(3))

import pandas as pd

#Create a list of document lables for row and column indices
doc_labels = [f'Document {i+1}' for i in range(len(documents))]

# Create a DataFrame for the similarity matrix
similarity_df = pd.DataFrame(
    similarity_matrix,
    index = doc_labels,
    columns = doc_labels
)

print('=== Cosine Similarity Table ===')
display(similarity_df.round(4))
```

#Using cosine similarity and TF-IDF

```python
#STEP 1. Import libraries
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# STEP 2. Develop the Knowledge base
questions = [
    "What are your office hours?",
    "Where is your office located?",
    "How can I contact support?",
    "What services do you offer?",
    "How much does the service cost?"
]

answers = [
    "Our office is open from 8 AM to 5 PM.",
    "Our office is located in Manila.",
    "You can contact support at support@example.com.",
    "We offer consultation and technical support.",
    "Please contact us for the latest pricing."
]

#STEP 3. Convert questions into numerical vectors
vectorizer = TfidfVectorizer()
question_vectors = vectorizer.fit_transform(questions)

def get_response(user_input):
  user_vector = vectorizer.transform([user_input])

  #STEP 4. Compare user input against stored questions
  similarities = cosine_similarity(
      user_vector, question_vectors
  ).flatten()

  best_match = similarities.argmax()
  confidence = similarities[best_match]

  # STEP 5. Reject weak matches
  if confidence < 0.2:
    return "Sorry, I don't know the answer to that."

  return answers[best_match]

print('FAQ Chatbot (type \'exit\' to stop)')

while True:
  user_input = input('You: ')
  if user_input.lower() == 'exit':
    print('Thank and Goodbye!')
    break

  print('Chatbot: ', get_response(user_input))
```
