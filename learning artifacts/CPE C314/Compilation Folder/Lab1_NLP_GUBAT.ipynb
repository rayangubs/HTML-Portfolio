# What is Regex?

Regex stands for Regular Expressions.
It’s a pattern-matching language used to search, match, and manipulate text.

    Think of it like a "super-powerful find & replace tool."
Instead of just looking for "dog", regex can look for:

- Any digit (\d)

- Any word (\w+)

- An email (\w+@\w+\.\w+)

- A phone number (\d{3}-\d{3}-\d{4})

Why use Regex?

    - Validation → Check if input is a valid email, phone, zip code
    - Search → Find patterns in text (all numbers, dates, hashtags, etc.)
    - Text cleaning → Remove punctuation, extract words, filter unwanted stuff

```python
import re

text = "My email is test123@example.com and phone is 123-456-7890."

# 1. Find all digits
print(re.findall(r'\d+', text))


# 2. Find all words
print(re.findall(r'\w+', text))


# 3. Find email pattern
print(re.findall(r'\w+@\w+.\w+', text))


# 4. Find phone number pattern
print(re.findall(r'\d+-\d+-\d+', text))
print(re.findall(r'\d{3}-\d{3}-\d{4}', text))
```

# Regex for NLP Preprocessing

```python
import re

text = "Check out https://example.com! Email me at test123@example.com. #NLP is amazing 2025!!!"
```

```python
#1. Lower casing
convert_lower = text.lower()
print(convert_lower)
```

```python
#2. Remove Punctuation
no_punct = re.sub(r'[^\w\s]', '', text)
print(no_punct)
```

```python
#3. Remove Numbers
no_num = re.sub(r'\d', '', text)
print(no_num)
```

```python
#4. Extract URLs
url = re.findall(r'\w+://\w+.\w+', text)
print(url)
```

```python
#5. Extract Emails
email = re.findall(r'\w+@\w+.\w+', text)
print(email)
sep_email = re.findall(r'\w+', text)
print(sep_email)
```

```python
#6. Extract Hashtags
hash = re.findall(r'#\w+', text)
print(hash)
```

```python
#7. Extract Words Only (Tokenization)
words = re.findall(r'\b\w+\b', convert_lower)
print(words)
```

```python
#8. Remove Emojis / Non-ASCII Characters
new = re.sub(r"[^\x00-\x7F]+", "", text)
print(new)
```

# Summary

- Regex helps with text cleaning in NLP:

- Remove punctuation → **re.sub(r"[^\w\s]", "", text)**

- Remove numbers → **re.sub(r"\d+", "", text)**

- Extract URLs → **re.findall(r"http\S+", text)**

- Extract emails → **re.findall(r"...", text)**

- Extract hashtags → **re.findall(r"#\w+", text)**

- Tokenize text → **re.findall(r"\b\w+\b", text)**

- Remove emojis/non-ASCII → **re.sub(r"[^\x00-\x7F]+", "", text)**

```python
#Another approach to stopwords

import re

STOPWORDS = {"is", "am", "are", "the", "a", "an", "and", "at", "in", "on", "for", "of",
             "to", "it", "this"}

def remove_stopwords(text, STOPWORDS):
  #Build regex pattern from stopword list
  pattern = r"\b(" + "|".join(STOPWORDS) + r")\b"
  print(pattern)
  # Replace stopwords with empty string
  cleaned = re.sub(pattern, "", text, flags=re.IGNORECASE)
  print(cleaned)
  # Remove extra spaces
  cleaned = re.sub(r"\s+", " ", cleaned).strip()
  print(cleaned)
  return cleaned

# ---------------- Example ----------------
text = "This is an example of a sentence in the NLP challenge."
print("Original:", text)
print("No stopwords:", remove_stopwords(text, STOPWORDS))
```

# How it works:

- \b → word boundary ensures whole words only.

- Without it, "this" would remove "thistle".

- (" + "|".join(STOPWORDS) + ") → builds a regex group like \b(is|am|are|the|a)\b.

- re.sub() → replaces stopwords with "".

- Clean up extra spaces.

```python
import re

#Sentence Input: Welcome to the TechHub portal, which was updated on 12 October 2025. Our team has successfully processed over 450 user accounts this week to improve network security. If you encounter any login issues or system bugs, please contact our support staff immediately at helpdesk247@techhub-solutions.org for assistance. Your security ticket will be generated automatically, and an agent will reply within 15 minutes.

text = input("Enter your Paragraph: \n")

#a. Convert to lowercase
lower_txt = text.lower()
print("\nLower Cased:")
print(lower_txt)

#b. Remove punctuation marks
no_punct = re.sub(r'[^\w\s]', '', lower_txt)
print("\nPunctuations Removed:")
print(no_punct)

#c. Remove numbers
no_num = re.sub(r'[\d]', '', no_punct)
print("\nNumbers Removed: ")
print(no_num)

#d. Remove all stopwords
STOPWORDS = {"is", "am", "are", "the", "a", "an", "and", "at", "in", "on", "for", "of",
             "to", "it", "this"}

#d.a Build regex pattern from stopword list
pattern = r"\b(" + "|".join(STOPWORDS) + r")\b"

#d.b Replace stopwords with empty string
cleaned = re.sub(pattern, "", no_num, flags=re.IGNORECASE)

#d.c Remove extra spaces
cleaned = re.sub(r"\s+", " ", cleaned).strip()

print("\nRemoved Stopwords: ")
print(cleaned)

#e. Tokenize the paragraph
final = re.findall(r'\b\w+\b', cleaned)
print("\nTokenized Paragraph: ")
print(final)
```

```python
import re

#https://facebook.com is my most favorite site - guaranteeeeed to make you soooo goooood - next to https://youtube.com. My user in facebook is @rayan.gubs and in youtube, it's @rayangubat. I like to follow the trends of #BINI and #BINI_Signals. Weird characters are é, à, ü, ñ. My number is 09123456789 and I like to add a bit to my username, like gubat123, or gubs@123__

email_input = input("Enter your Paragraph: \n")

#Text Preprocessing - use Regex

# 1.  Remove URLs (http...)
url_removed = re.sub(r'http\S+', '', email_input)
print('\nURL Removed: ')
print(url_removed)

# 2.  Remove mentions (@username)
mention_removed = re.sub(r'@\S+', '', url_removed)
print('\nMentions (@username) Removed: ')
print(mention_removed)

# 3.  Remove hashtags symbol # but keep the word (e.g., #happy → happy)
hash_removed = re.sub(r'#', '', mention_removed)
print('\nHash (#) Symbol Removed: ')
print(hash_removed)

# 4.  Remove emojis & non-ASCII characters
spec_removed = re.sub(r'[^\x00-\x7F]+', '', hash_removed)
print('\nEmojis and non-ASCII characters Symbol Removed: ')
print(spec_removed)

# 5.  Remove numbers and punctuation
punc_removed = re.sub(r'[^\w\s]', '', spec_removed)
np_removed = re.sub(r'\d+', '', punc_removed)
print('\nNumbers and Punctuations Removed: ')
print(np_removed)

# 6.  Normalize repeated characters (e.g., soooo → soo)
pattern_norm = r'(\w)\1+'
normalize = re.sub(pattern_norm, r'\1\1', np_removed)
print('\nNormalized Text: ')
print(normalize)

# 7.  Convert to lowercase
lowered_text = normalize.lower()
print('\nLower Cased Text: ')
print(lowered_text)

# 8.  Tokenize text into words
tokenized_word = re.findall(r'\b\w+\b', lowered_text)
print("\nTokenized Text: ")
print(tokenized_word)

# 9.  Stopword Removal
STOPWORDS = {"is", "am", "are", "the", "a", "an", "and", "at", "in", "on", "for", "of",
             "to", "it", "this"}

cleaned = [word for word in tokenized_word if word not in STOPWORDS]

print("\nStopword Removal: ")
print(cleaned)

# 10. Remove common stopwords (you may define a list manually).
CUSTOM_SW = {
    "i", "me", "my", "myself", "we", "our", "ours", "ourselves", "you", "your", "yours",
    "yourself", "yourselves", "he", "him", "his", "himself", "she", "her", "hers",
    "herself", "they", "them", "their", "theirs", "themselves", "what", "which", "who",
    "whom", "was", "were", "be", "been", "being", "have", "has", "had", "having", "do",
    "does", "did", "doing", "will", "would", "shall", "should", "can", "could", "may",
    "might", "must", "but", "if", "or", "because", "as", "until", "while", "with",
    "about", "against", "between", "into", "through", "during", "before", "after",
    "above", "below", "from", "up", "down", "out", "off", "over", "under", "again",
    "further", "then", "once", "here", "there", "when", "where", "why", "how", "all",
    "any", "both", "each", "few", "more", "most", "other", "some", "such", "no", "nor",
    "not", "only", "own", "same", "so", "than", "too", "very", "just", "now", "im",
    "dont", "cant", "wont", "isnt", "arent", "didnt", "doesnt", "wouldnt", "couldnt",
    "shouldnt", "thats", "its", "theyre", "youre", "weve", "youve"
}

final_text = [word for word in cleaned if word not in CUSTOM_SW]

print("\nFinal Removal and Final List: ")
print(f'{final_text}')
```
